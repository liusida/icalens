```python
from icalens import ICALens

lens = ICALens.from_pretrained("sida/icalens-qwen3.5-2b-ultrachat-1m")
messages = [{
    "role": "user",
    "content": "If you had to pick one, what is the most interesting science? Be brief.",
}]
```

```python
baseline = lens.generate(
    messages,
    max_new_tokens=16,
)
print("Baseline:", baseline)
```

```python
concept_result = lens.analyze(
    "Quantum computing uses quantum mechanics, qubits, superposition, and entanglement; "
    "neuroscience studies neurons, synapses, cognition, and the brain.",
    layer=5,
)
concept_result
```

```python
baseline_messages = [
    *messages,
    {"role": "assistant", "content": baseline},
]
baseline_result = lens.analyze(baseline_messages, layer=5)
baseline_result
```

```python
steered = lens.generate(
    messages,
    layer=5,
    clamp=(188, -20.0),
    max_new_tokens=16,
)
print("Steered:", steered)
```

```python
steered_messages = [
    *messages,
    {"role": "assistant", "content": steered},
]
steered_result = lens.analyze(steered_messages, layer=5)
steered_result
```

```python
for target in (-5.0, -10.0, -15.0, -17.0, -20.0, -25.0):
    response = lens.generate(
        messages,
        layer=5,
        clamp=(188, target),
        max_new_tokens=16,
    )
    print(target, response.splitlines()[0])
```
