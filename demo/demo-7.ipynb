```python
from icalens import ICALens

lens = ICALens.from_pretrained(
    "sida/icalens-qwen3.5-2b-ultrachat-1m"
)

query = "biology"

for layer in lens.available_layers:
    for component in range(lens.metadata["layers"][str(layer)]["n_components"]):
        profile = lens.component_profile(layer=layer, component=component)

        matches = []

        # Tokens observed in high-energy dataset examples
        for sign in ("positive", "negative"):
            for item in profile["examples"][sign]["tokens"]:
                if query.lower() in item["text"].lower():
                    matches.append(
                        f"examples/{sign}: {item['text']!r}, count={item['count']}"
                    )

        # Tokens associated through final norm and unembedding
        for sign in ("positive", "negative"):
            for rank_type in ("top_tokens", "bottom_tokens"):
                for rank, item in enumerate(
                    profile["logit_lens"][sign][rank_type], start=1
                ):
                    if query.lower() in item["text"].lower():
                        matches.append(
                            f"logit_lens/{sign}/{rank_type}: "
                            f"{item['text']!r}, rank={rank}, logit={item['logit']:+.3f}"
                        )

        if matches:
            print(
                f"Layer {layer}, C{component}, "
                f"dominant sign={profile['dominant_sign']}"
            )
            for match in matches:
                print(f"  {match}")
```

```python
from pprint import pprint

from icalens import ICALens

lens = ICALens.from_pretrained(
    "sida/icalens-qwen3.5-2b-ultrachat-1m"
)

profile = lens.component_profile(layer=21, component=644)

print("Dominant sign:", profile["dominant_sign"])
print("Sign statistics:")
pprint(profile["sign_statistics"])

print("\nNegative top-example token counts:")
pprint(profile["examples"]["negative"]["tokens"])

print("\nNegative high-energy occurrences:")
for example in profile["examples"]["negative"]["occurrences"]:
    print(
        f"energy={example['energy']:.2%}  "
        f"score={example['score']:+.3f}  "
        f"token={example['text']!r}  "
        f"position={example['position']}"
    )
    print(f"  context: {example['context']!r}")
```
